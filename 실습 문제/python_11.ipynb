## **9.8, 9.9** 다음과 같은 기능을 가지는 Counter 클래스를 정의하고 인스턴스를 구현하여라.

속성

* 이 클래스는 `_number`라는 속성 값을 가짐.
* `_number` 속성은 number를 매개 변수로 가지는 `__init__(self, number)`  생성자에서 초기화 . number의 디폴트 매개변수 값은 0임. number가 100 이상이 되거나 -1 이하가 되면 0 값을 초기 값으로 가짐.

메소드

* `reset()`: `_number`를 0으로 초기화
* `inc()`: `_number`를 1 증가. `_number`가 100 이상이 되면 0으로 초기화 됨.
* `dec()`:  `_number`를 1 감소. `_number`가 -1 이하가 되면 0 값을 가짐.
- `__str__()` : 문자열 `'C(n)'`를 반환
- c1 + c2 연산을 통해서 c1의 number와  c2의 number의 합을 속성으로 가지는 Counter 객체를 반환한다. 이때 number가 100 이상이 되면 0으로 초기화 됨.
- c1 - c2 연산을 통해서 c1의 number와  c2의 number의 차를 속성으로 가지는 Counter 객체를 반환한다. 이때 number가 -1 이하가 되면 0 값을 가짐.

### 이 인스턴스를 사용하여 다음 예제 1, 예제 2, 예제 3과 같은 코드를 작성하고 그 결과가 아래와 같이 나오도록 하시오.

#### **예제 1)**

```
c1 = Counter(10)
c1.inc()
print('c1 =', c1)
c2 = Counter(99)
c2.inc()
print('c2 =', c2)
```
실행 결과



```
c1 = C(11)
c2 = C(0)
```



#### **예제 2)**


```
c1 = Counter()
c1.inc()
c1.inc()
print('c1 =', c1)
c1.reset()
print('c1 =', c1)
c1.dec()
print('c1 =', c1)
```

실행 결과


```
c1 = C(2)
c1 = C(0)
c1 = C(0)
```






#### **예제 3)**

```
c1 = Counter(90)
c2 = Counter(20)
print('c1 + c2 =', c1 + c2)
print('c1 - c2 =', c1 - c2)
```
실행 결과


```
c1 + c2 = C(0)
c1 - c2 = C(70)
```

## 2. Elevator 클래스를 작성한 후, Elevator 객체를 생성하여 아래와 같이 테스트하시오.

* 속성(인스턴스 변수, 멤버 변수) : **private**
  * `_floor` : 현재 층을 저장. 단, -2층에서 10층까지의 층만 유효하며, 0층은 로비를 의미한다.
  * `_is_open` : 문 열림 상태를 저장. 문이 열린 경우 True, 문이 닫힌 경우 False.

* 동작(메소드)
  * `__init__()` : 생성자. 엘레베이터의 속성(현재 층과 문 상태)을 매개 변수로 받아 초기화한다. 매개 변수는 로비, 닫힘을 디폴트 값으로 가진다. 유효하지 않은 층수로 초기화를 시도할 경우 로비층으로 초기화할 것.
  *  `close()`: 문이 열린 경우 닫는다.
  *   `open()`: 문이 닫힌 경우 연다.
  * `move(destination)` : 문을 닫은 후 도착 층(destination)으로 이동하며, 그 과정을 보여(출력)준다. 도착 층으로 이동했다면 문을 연다. 단, 도착 층이 유효하지 않은 값인 경우 '유효한 층수를 입력해주세요.'라는 문장을 출력한다.
