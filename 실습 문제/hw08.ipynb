## 3.문장을 매개변수로 받아 문장에 포함된 단어들을   반환하는 함수 `get_words()`를 작성하시오. 그 후, 사용자로부터 문장을 입력받아 문장에 포함된  단어들을  출력하시오.

 `get_words()`
  
  - 단, 문장에 포함된 단어는 공백 및 구두점으로 구분한다.


 `get_words()` 함수 테스트

 - 사용자로부터 입력받은 문장을 인자로 하여 `get_words()` 함수를 호출한다.
 -  `get_words()` 함수의 반환값을 활용하여, 문장에 포함된 단어를 `,`로 구분하여 출력할 것.
 - 문장에 포함된 단어가 없다면(즉, 반환값이 없거나 비었다면), '문장에 포함된 단어는 없습니다.'라는 문장을 출력할 것.


 ---
 실행 결과

```
문장: Hello, python! hello, world. Python is fun.
문장에 포함된 단어: Hello, python, hello, world, Python, is, fun
```



 실행 결과

```
문장: !!! .. ,
문장에 포함된 단어가 없습니다.
```

### 3.1 이번에는 문장을 매개변수로 받아 문장에 포함된 단어들을  중복없이 반환하는 함수 `get_distinct_words()`를 작성하시오. 그 후, 사용자로부터 문장을 입력받아 문장에 포함된  단어들을 중복없이 출력하시오.

 `get_distinct_words()`
  
  - **중복 제거**:  대소문자를 무시하고 중복을 판단한다.
  
  - **원문의 대소문자를 유지**하여 반환할 것. <- 이 부분 구현은 난이도가 높습니다.




 `get_distinct_words()` 함수 테스트

 - 사용자로부터 입력받은 문장을 인자로 두고 `get_distinct_words()` 함수를 호출한다.
 -  `get_distinct_words()` 함수의 반환값을 활용하여, 문장에 포함된 단어를 `,`로 구분하여 출력할 것.
 - 문장에 포함된 단어가 없다면, '문장에 포함된 단어는 없습니다.'라는 문장을 출력할 것.


 ---
 실행 결과

```
문장: Hello, python! hello, world. Python is fun.
문장에 포함된 단어: Hello, python, world, is, fun #Python, python 중 어느 것을 남겨도 무방
```



 실행 결과

```
문장: !!! .. ,
문장에 포함된 단어가 없습니다.
```

## 4. 사용자로부터 문장과 검색할 단어를 차례로 입력받아, 문장에서 해당 단어 검색 횟수를 출력하시오.
 - **대소문자 무시하고 비교**할 것.
 - 단어의 일부가 아닌, 독립된 단어로 존재하는 경우에만 횟수에 포함할 것.
   - 'Oh Foo!'에서는 'FOO'가 검색됨.
   - 'Fool's day.'에서는 'FOO'가 검색되지 않음: 'Fool'의 일부이기 때문
 - 단어는 구두점 혹은 공백으로 구분할 것.


---
실행 결과

```
문장: This is a foo situation. Oh Foo! I can't believe this foo happened. Fool's day.
검색할 단어: FOo
FOo 검색 횟수: 3 #결과 출력시 키워드는 사용자가 입력한 대소문자를 유지하여 출력할 것. 즉, foo 혹은 FOO로 바꿔 출력하면 안 됨.
```

### 4.1 [조건 추가]  검색 횟수 뿐 아니라, 검색된 단어도 출력하시오.
- 결과 출력시, 아래와 같이 `'원문'에서 '키워드' 검색 횟수: 횟수 (검색된 단어)`의 형식으로 출력할 것.
- 키워드가 한번도 검색되지 않은 경우 `'원문'에서 '키워드'가 검색되지 않았습니다.`의 형식으로 출력할 것.
 - 단, 검색된 단어 출력시 사용자가 입력한 대소문자를 유지하여 출력할 것.
---

실행 결과
```
문장: This is a foo situation. Oh Foo! I can't believe this foo happened. Fool's day.
검색할 단어: foO
'This is a foo situation. Oh Foo! I can't believe this foo happened. Fool's day.' 문장에서 'foO' 검색 횟수: 3 (foo, Foo, foo)
```

### 애너그램(anagram)이란 문자의 순서를 섞어 만든 새로운 단어다.







 `gen_anagram()` 함수

  - 문자열을 매개변수로 받아 애너그램을 반환한다.


 `gen_anagram()` 함수 테스트

 - 사용자가 단어 한 개를 입력하지 않은 경우 오류 메시지를 출력하고 다시 입력받을 것.

```
단어: python is fun
[입력오류] 단어 한 개를 입력해주세요.
단어:
[입력오류] 단어 한 개를 입력해주세요.
단어: python
python의 애너그램: yntohp
```
