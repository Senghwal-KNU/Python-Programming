### 5.7.2 세 수를 매개변수로 받아 정렬하여 **출력**하는 'sort()` 함수를 추가 작성하고, 이를 활용하여 아래와 같이 출력하시오.
 - 사용자로부터 입력받은 세 수를 함수의 인자로 전달할 것.
 - 파이썬 내장 함수를 사용하지 말 것.
---
테스트 코드
```
sort(a, b, c)
print(f'의 평균값은 {mean3(a, b, c):.1f}')
sort(a, b, c)
print(f'의 최댓값은 {max3(a, b, c):.1f}')
sort(a, b, c)
print(f'의 최솟값은 {min3(a, b, c):.1f}')
```


실행 결과
```
세 수를 입력하시오: 5.4 9.1 -2.7
-2.7, 5.4, 9.1의 평균값은 3.9
-2.7, 5.4, 9.1의 최댓값은 9.1
-2.7, 5.4, 9.1의 최솟값은 -2.7
```
