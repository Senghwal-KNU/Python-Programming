## 1. 문구점에서 공책(￦1,000), 연필(￦500), 지우개(￦500), 볼펜(￦1,000)을 판매한다. 이 문구점의 메뉴판을 딕셔너리로 작성한 후 아래와 같이 출력하시오.

---
실행 결과
```
품목    가격
***************
공책    1,000원
연필      500원
지우개    500원
볼펜    1,000원
```


### 1. 문구점의 취급 품목과 그 개수를 출력하시오.

---
실행 결과
```
취급 품목(총 4개): 공책, 연필, 지우개, 볼펜
```

### 2. 사용자에게 상품명을 반복하여 입력받아, 장바구니에 추가하는 프로그램을 작성하시오.
  - 장바구니는 딕셔너리로 구현한다. 이 때, key는 사용자가 장바구니에 추가한 상품명, value는 주문 수량을 의미한다.  
 - 사용자가 '나가기'를 입력하면 물품 담기를 종료한다.

---
실행 예시
```
상품: 공책
공책을 장바구니에 담았습니다.
상품: 공책
공책을 장바구니에 담았습니다.
상품: 연필
연필을 장바구니에 담았습니다.
상품: 지우개
지우개을 장바구니에 담았습니다.
상품: 테이프
테이프은 취급하지 않습니다.
상품: 나가기
```

위와 같이 실행하면 `cart` 는 `{'공책': 2, '연필': 1, '지우개': 1}`입니다.

---

 - idea sketch
  
    1. 반복문 작성 전, 빈 딕셔너리 `cart`를 생성한다.
    2. 반복문 안에서,
     -  사용자가 입력한 물품이 `cart`에 이미 있다면 수량 추가
     -  사용자가 입력한 물품이 `cart`에 없다면 수량 1로 설정
     

### 3. 장바구니에 담긴 내역과 총 주문 금액을 출력하시오.
---
실행 예시
```
장바구니 내역
공책(주문 수량: 2)
연필(주문 수량: 1)
지우개(주문 수량: 1)
주문 금액: 3,000원
```
