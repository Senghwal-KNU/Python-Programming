# **Mutable and Immutable**

##1. `is`연산과 `id()` 함수(feat. mutable/immutable)



###Mutable (가변 객체): 값을 변경할 수 있는 객체
- 예: list, set, dict, 사용자 정의 클래스(Class)의 객체 등



```
dct1 = {'Monday':1, 'Tuesday':2}
dct2 = {'Tuesday':2, 'Monday':1}

print('dct1==dct2:', dct1 == dct2)
print('dct1 is dct2:', dct1 is dct2)  
```



###Immutable (불변 객체): 값을 변경할 수 없는 객체
 - 예: int, float, str, tuple, range() 등




```
x = 1
y = 1
print('x==y:', x==y)        
print('x is y:', x is y)    
```

 - 단, `lst`는 `['dog', 'cat', 'tiger', 'lion']`이다.

 - `rrot()`: 반환값 없음.


---
실행 예시

```
n: 10
오른쪽으로 10칸 로테이션 후: ['tiger', 'lion', 'dog', 'cat']
```
