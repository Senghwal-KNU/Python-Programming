### 1. 사용자로부터 원의 반지름(실수)를 입력받아 원의 둘레의 길이와 넓이를 출력하시오.
---
실행 결과

```
원의 반지름: 2.7
반지름이 2.7인 원의 넓이는 22.89, 둘레의 길이는 16.96입니다.
```

### 2. 사용자로부터 두 개의 정수 `a`, `b`를 입력받아 아래와 같이 사칙연산 결과를 출력하시오.


---
실행 결과
```
정수 a: 7
정수 b: 5

7 + 5 = 12
7 - 5 = 2
7 * 5 = 35
7 / 5 = 1.4
```
