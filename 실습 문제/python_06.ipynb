---
```
>>> s = 'Birthday'
>>> s[:5]
1) _____________   
>>> s[5:]
2) _____________   
>>> s [1:-1]
3) _____________   
>>> s [::-1]
4) _____________   
>>> 'day' in s
5) _____________   
>>> 'birth' in s
6) _____________   
>>> 'Birth' in s
7) _____________   
>>> 'Birth' not in s
8) _____________   
```
